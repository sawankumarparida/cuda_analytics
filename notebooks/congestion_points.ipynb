```python
import cudf
from keplergl import KeplerGl

print("🚀 Loading 500,000 rows directly into GPU VRAM...")
# 1. Load data using RAPIDS cuDF instead of Pandas
# This utilizes your CUDA cores to load and process the data instantly!
gdf = cudf.read_csv('spatial_telemetry.csv')

# Let's do some quick GPU-accelerated Data Engineering to prove it works
severe_congestion = gdf[gdf['velocity_kmh'] < 30]
print(f"⚠️ GPU identified {len(severe_congestion):,} severe congestion points in milliseconds.")

print("🗺️ Initializing WebGL Map...")
# 2. Create the Kepler WebGL map instance
map_1 = KeplerGl(height=700)

# 3. Feed the data to the map
# We use .to_pandas() at the very last step just to hand the GPU-processed data 
# to the web browser for rendering.
map_1.add_data(data=gdf.to_pandas(), name='Telemetry_Stream')

# 4. Display the interactive map in the notebook
map_1
```
